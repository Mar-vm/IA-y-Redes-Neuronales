## PASO 4 — Aplicación de mejoras básicas
**Valor: 2% | Tiempo estimado: 1 hora**

> 🔧 **Experimenta:** Modifica los valores de ALPHA, BETA, KERNEL_GAUSS y SIGMA y observa cómo cambia la imagen.

````python
# ─── PARÁMETROS — modifica y observa los cambios ─────────────────────────────
ALPHA = 1.5      # Contraste: < 1 reduce, = 1 sin cambio, > 1 aumenta
BETA  = 30       # Brillo   : negativo oscurece, positivo ilumina
KERNEL_GAUSS = (5, 5)  # Tamaño del kernel gaussiano (debe ser impar × impar)
SIGMA = 1.0      # Desviación estándar del gaussiano (suavidad)

# ── 1) Ajuste de brillo y contraste ──────────────────────────────────────────
#    Fórmula: nueva_img[x,y] = alpha × img[x,y] + beta
#    cv2.convertScaleAbs garantiza valores en rango [0, 255]
img_brillo_contraste = cv2.convertScaleAbs(img_gray, alpha=ALPHA, beta=BETA)

# ── 2) Suavizado gaussiano ────────────────────────────────────────────────────
#    Reduce ruido mediante un promedio ponderado con distribución gaussiana
#    Útil antes de detección de bordes para evitar falsos positivos por ruido
img_suavizada = cv2.GaussianBlur(img_gray, KERNEL_GAUSS, SIGMA)

# ── 3) Ecualización de histograma ─────────────────────────────────────────────
#    Redistribuye los niveles de intensidad para aprovechar todo el rango [0, 255]
#    Mejora el contraste global, especialmente en imágenes oscuras o lavadas
img_ecualizada = cv2.equalizeHist(img_gray)

# ── 4) Realce de bordes con Sobel ─────────────────────────────────────────────
#    Calcula el gradiente de intensidad: detecta cambios bruscos (bordes)
#    Clave para entender las capas convolucionales de una CNN (Tema 3)
sobelx    = cv2.Sobel(img_gray, cv2.CV_64F, 1, 0, ksize=3)   # Gradiente horizontal
sobely    = cv2.Sobel(img_gray, cv2.CV_64F, 0, 1, ksize=3)   # Gradiente vertical
img_bordes = np.uint8(np.clip(cv2.magnitude(sobelx, sobely), 0, 255))  # Magnitud

# ── Visualización comparativa ─────────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle('Paso 4 — Mejoras básicas aplicadas a la imagen', fontsize=14, fontweight='bold')

datos = [
    (img_gray,             'Original (grises)'),
    (img_brillo_contraste, f'Brillo/Contraste\nα={ALPHA}, β={BETA}'),
    (img_suavizada,        f'Suavizado Gaussiano\n{KERNEL_GAUSS}, σ={SIGMA}'),
    (img_ecualizada,       'Ecualización\nde histograma'),
    (img_bordes,           'Realce de bordes\n(Sobel)'),
]
for ax, (img, tit) in zip(axes.ravel(), datos):
    ax.imshow(img, cmap='gray')
    ax.set_title(tit, fontsize=10)
    ax.axis('off')
axes[1, 2].axis('off')

plt.tight_layout()
plt.savefig('paso4_mejoras.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')

print('Parámetros utilizados:')
print(f'  alpha (contraste)  : {ALPHA}')
print(f'  beta (brillo)      : {BETA}')
print(f'  kernel gaussiano   : {KERNEL_GAUSS}')
print(f'  sigma              : {SIGMA}')
print(f'  ksize Sobel        : 3×3')
print('Guardado: paso4_mejoras.png')
````

### Variante B — Filtro de mediana vs gaussiano (ruido sal y pimienta)

````python
# ══ VARIANTE B — Filtro de mediana vs gaussiano (ruido sal y pimienta) ═══════
# Esta variante demuestra cuándo el filtro de mediana es superior al gaussiano.
# El ruido sal y pimienta son píxeles aleatorios en blanco (sal) o negro (pimienta).

# Agregar ruido sal y pimienta artificialmente al 2% de los píxeles
np.random.seed(42)
img_ruidosa = img_gray.copy().astype(np.float32)
n_px = int(0.02 * img_gray.size)

coords_sal = [np.random.randint(0, d, n_px) for d in img_gray.shape]
img_ruidosa[coords_sal[0], coords_sal[1]] = 255   # sal (blanco)

coords_pim = [np.random.randint(0, d, n_px) for d in img_gray.shape]
img_ruidosa[coords_pim[0], coords_pim[1]] = 0     # pimienta (negro)
img_ruidosa = np.uint8(img_ruidosa)

img_mediana   = cv2.medianBlur(img_ruidosa, 5)              # Filtro de mediana 5×5
img_gaussiana = cv2.GaussianBlur(img_ruidosa, (5, 5), 1.0)  # Gaussiano 5×5

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Variante B — Mediana vs Gaussiano para ruido sal y pimienta',
             fontsize=12, fontweight='bold')
axes[0].imshow(img_ruidosa,   cmap='gray')
axes[0].set_title('Con ruido S&P (2%)')
axes[0].axis('off')
axes[1].imshow(img_gaussiana, cmap='gray')
axes[1].set_title('Filtro Gaussiano 5×5\n(difumina bordes)')
axes[1].axis('off')
axes[2].imshow(img_mediana,   cmap='gray')
axes[2].set_title('Filtro Mediana 5×5\n(preserva bordes)')
axes[2].axis('off')

plt.tight_layout()
plt.savefig('paso6_variante_b.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')
print('Observación: el filtro de mediana elimina el ruido S&P preservando mejor los bordes.')
print('Guardado: paso6_variante_b.png')
````

### Variante D — Umbralización (segmentación binaria)

> **Concepto:** La umbralización convierte una imagen en escala de grises a blanco/negro según un umbral de intensidad. Es el primer paso en muchos sistemas de segmentación.
>
> - **Global fijo:** todos los píxeles > umbral → blanco, resto → negro.
> - **Otsu:** calcula automáticamente el mejor umbral minimizando la varianza intraclase.
> - **Adaptativo:** calcula un umbral diferente para cada región local (útil con iluminación irregular).

````python
# ══ VARIANTE D — Umbralización ════════════════════════════════════════════════
# Segmenta la imagen en regiones binarias (blanco/negro).
# Fundamental para detectar objetos, textos o regiones de interés.

# ─── PARÁMETRO — prueba valores entre 50 y 200 ───────────────────────────────
UMBRAL_GLOBAL = 127

# 1) Umbral fijo global
_, img_thresh_global = cv2.threshold(
    img_gray, UMBRAL_GLOBAL, 255, cv2.THRESH_BINARY
)

# 2) Umbral de Otsu (automático — no requiere parámetro manual)
otsu_val, img_thresh_otsu = cv2.threshold(
    img_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# 3) Umbral adaptativo gaussiano (ajusta por regiones de 11×11 px)
img_thresh_adapt = cv2.adaptiveThreshold(
    img_gray, 255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    blockSize=11,   # Tamaño de la vecindad (debe ser impar)
    C=2             # Constante que se resta de la media local
)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
fig.suptitle('Variante D — Umbralización (segmentación binaria)',
             fontsize=12, fontweight='bold')

axes[0].imshow(img_gray,          cmap='gray')
axes[0].set_title('Original (grises)'); axes[0].axis('off')

axes[1].imshow(img_thresh_global, cmap='gray')
axes[1].set_title(f'Umbral global fijo\nT={UMBRAL_GLOBAL}')
axes[1].axis('off')

axes[2].imshow(img_thresh_otsu,   cmap='gray')
axes[2].set_title(f'Otsu (automático)\nT calculado = {otsu_val:.0f}')
axes[2].axis('off')

axes[3].imshow(img_thresh_adapt,  cmap='gray')
axes[3].set_title('Adaptativo gaussiano\n(por regiones locales)')
axes[3].axis('off')

plt.tight_layout()
plt.savefig('paso6_variante_d.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')

print(f'Umbral global usado : {UMBRAL_GLOBAL}')
print(f'Umbral calculado por Otsu: {otsu_val:.1f}')
print('Guardado: paso6_variante_d.png')
print()
print('Pregunta guía: ¿En qué condiciones Otsu falla y el umbral adaptativo es mejor?')
````

### Variante E — Detector de bordes Canny

> **Concepto:** Canny es un detector de bordes multi-etapa que aplica:
> 1. Suavizado gaussiano (elimina ruido)
> 2. Gradiente de Sobel (encuentra cambios de intensidad)
> 3. Supresión de no-máximos (adelgaza los bordes a 1 px de grosor)
> 4. Histéresis con dos umbrales (elimina bordes débiles no conectados)
>
> Produce bordes más limpios y continuos que Sobel solo.

````python
# ══ VARIANTE E — Detector de bordes Canny ════════════════════════════════════
# Canny produce bordes más definidos que Sobel al incluir histéresis y supresión
# de no-máximos. Es el detector de bordes más usado en visión por computadora.

# ─── PARÁMETROS — threshold1 debe ser < threshold2 ──────────────────────────
# Regla práctica: threshold2 ≈ 2–3× threshold1
CANNY_T1_SUAVE    = 30    # Umbral bajo  → más bordes detectados
CANNY_T2_SUAVE    = 90
CANNY_T1_ESTRICTO = 100   # Umbral alto  → solo bordes fuertes
CANNY_T2_ESTRICTO = 200

# Suavizar antes de Canny reduce falsos positivos por ruido
img_pre_canny = cv2.GaussianBlur(img_gray, (5, 5), 1.0)

img_canny_suave    = cv2.Canny(img_pre_canny, CANNY_T1_SUAVE,    CANNY_T2_SUAVE)
img_canny_estricto = cv2.Canny(img_pre_canny, CANNY_T1_ESTRICTO, CANNY_T2_ESTRICTO)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
fig.suptitle('Variante E — Canny vs Sobel: detección de bordes',
             fontsize=12, fontweight='bold')

axes[0].imshow(img_gray,           cmap='gray')
axes[0].set_title('Original'); axes[0].axis('off')

axes[1].imshow(img_bordes,         cmap='gray')
axes[1].set_title('Sobel\n(del Paso 4)'); axes[1].axis('off')

axes[2].imshow(img_canny_suave,    cmap='gray')
axes[2].set_title(f'Canny (sensible)\nT1={CANNY_T1_SUAVE}, T2={CANNY_T2_SUAVE}')
axes[2].axis('off')

axes[3].imshow(img_canny_estricto, cmap='gray')
axes[3].set_title(f'Canny (estricto)\nT1={CANNY_T1_ESTRICTO}, T2={CANNY_T2_ESTRICTO}')
axes[3].axis('off')

plt.tight_layout()
plt.savefig('paso6_variante_e.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')

# Comparar cantidad de píxeles de borde detectados
px_sobel  = np.count_nonzero(img_bordes > 30)
px_suave  = np.count_nonzero(img_canny_suave)
px_strict = np.count_nonzero(img_canny_estricto)
print(f'Píxeles de borde detectados:')
print(f'  Sobel            : {px_sobel:,}')
print(f'  Canny sensible   : {px_suave:,}')
print(f'  Canny estricto   : {px_strict:,}')
print('Guardado: paso6_variante_e.png')
````

### Variante F — Espacio de color HSV

> **Concepto:** El modelo HSV (Hue, Saturation, Value) separa la información de color de la información de iluminación:
> - **H (Hue / Tono):** tipo de color (0–179 en OpenCV → rojo, verde, azul, etc.)
> - **S (Saturation / Saturación):** pureza del color (0 = gris, 255 = color puro)
> - **V (Value / Valor):** brillo (0 = negro, 255 = máximo brillo)
>
> El canal V es mucho más estable ante cambios de iluminación que cualquier canal RGB.

````python
# ══ VARIANTE F — Espacio de color HSV ════════════════════════════════════════
# HSV facilita la detección de objetos por color sin depender tanto de la iluminación.
# Muy usado en robótica, seguimiento de objetos y detección de piel.

img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
canal_h = img_hsv[:, :, 0]   # Tono     (0-179 en OpenCV)
canal_s = img_hsv[:, :, 1]   # Saturación (0-255)
canal_v = img_hsv[:, :, 2]   # Valor / Brillo (0-255)

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle('Variante F — Espacio de color HSV vs RGB',
             fontsize=13, fontweight='bold')

# Fila 1: canales RGB
for i, (canal, tit, cm) in enumerate(zip(
    [canal_r, canal_g, canal_b],
    ['Canal R (Rojo)', 'Canal G (Verde)', 'Canal B (Azul)'],
    ['Reds_r', 'Greens_r', 'Blues_r']
)):
    axes[0, i].imshow(canal, cmap=cm)
    axes[0, i].set_title(f'{tit}\nMedia: {canal.mean():.1f}')
    axes[0, i].axis('off')

# Fila 2: canales HSV
for i, (canal, tit, cm) in enumerate(zip(
    [canal_h, canal_s, canal_v],
    ['Canal H (Tono)', 'Canal S (Saturación)', 'Canal V (Valor/Brillo)'],
    ['hsv', 'YlOrRd', 'gray']
)):
    axes[1, i].imshow(canal, cmap=cm)
    axes[1, i].set_title(f'{tit}\nMedia: {canal.mean():.1f}')
    axes[1, i].axis('off')

plt.tight_layout()
plt.savefig('paso6_variante_f.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')

print('Estadísticas por canal HSV:')
print(f'  H (Tono)        — min: {canal_h.min()}, max: {canal_h.max()}, media: {canal_h.mean():.1f}')
print(f'  S (Saturación)  — min: {canal_s.min()}, max: {canal_s.max()}, media: {canal_s.mean():.1f}')
print(f'  V (Brillo)      — min: {canal_v.min()}, max: {canal_v.max()}, media: {canal_v.mean():.1f}')
print('Guardado: paso6_variante_f.png')
print()
print('Pregunta guía: ¿El canal V de HSV es similar a la imagen en escala de grises? ¿Por qué?')
````

### Variante G — Operaciones morfológicas (erosión y dilatación)

> **Concepto:** Las operaciones morfológicas actúan sobre la **forma** de las regiones en imágenes binarias:
> - **Erosión:** encoge las regiones blancas. Elimina ruido pequeño y separa objetos pegados.
> - **Dilatación:** agranda las regiones blancas. Rellena huecos y conecta objetos cercanos.
> - **Apertura** (erosión → dilatación): elimina ruido sin cambiar el tamaño general.
> - **Cierre** (dilatación → erosión): rellena huecos internos.
>
> Son fundamentales como paso de postprocesamiento después de umbralización o detección de bordes.

````python
# ══ VARIANTE G — Morfología: erosión, dilatación, apertura y cierre ══════════
# Se aplica sobre la imagen umbralizada (Otsu) del Paso D.
# Si no ejecutaste la Variante D, se calcula aquí automáticamente.

try:
    base_morfo = img_thresh_otsu.copy()
except NameError:
    _, base_morfo = cv2.threshold(img_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# ─── PARÁMETRO — prueba tamaños 3, 5, 7, 9 ──────────────────────────────────
TAM_KERNEL = 5
kernel = np.ones((TAM_KERNEL, TAM_KERNEL), np.uint8)

img_erosion  = cv2.erode(base_morfo,  kernel, iterations=1)
img_dilacion = cv2.dilate(base_morfo, kernel, iterations=1)
img_apertura = cv2.morphologyEx(base_morfo, cv2.MORPH_OPEN,  kernel)  # erosión → dilatación
img_cierre   = cv2.morphologyEx(base_morfo, cv2.MORPH_CLOSE, kernel)  # dilatación → erosión

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
fig.suptitle(f'Variante G — Morfología (kernel {TAM_KERNEL}×{TAM_KERNEL})',
             fontsize=12, fontweight='bold')

for ax, img, tit in zip(
    axes,
    [base_morfo, img_erosion, img_dilacion, img_apertura, img_cierre],
    ['Original (Otsu)', f'Erosión\n(encoge regiones)',
     f'Dilatación\n(agranda regiones)',
     f'Apertura\n(elimina ruido)',
     f'Cierre\n(rellena huecos)']
):
    ax.imshow(img, cmap='gray')
    px_blancos = np.count_nonzero(img)
    ax.set_title(f'{tit}\n{px_blancos:,} px blancos', fontsize=9)
    ax.axis('off')

plt.tight_layout()
plt.savefig('paso6_variante_g.png', dpi=150, bbox_inches='tight')
plt.show()
plt.close('all')

print(f'Kernel utilizado: {TAM_KERNEL}×{TAM_KERNEL}')
print(f'Px blancos — Original: {np.count_nonzero(base_morfo):,}')
print(f'Px blancos — Erosión : {np.count_nonzero(img_erosion):,}  '
      f'(reducción: {np.count_nonzero(base_morfo)-np.count_nonzero(img_erosion):,} px)')
print(f'Px blancos — Dilación: {np.count_nonzero(img_dilacion):,}  '
      f'(aumento  : {np.count_nonzero(img_dilacion)-np.count_nonzero(base_morfo):,} px)')
print('Guardado: paso6_variante_g.png')
print()
print('Pregunta guía: ¿Cuándo usarías apertura en lugar de erosión simple?')
````
